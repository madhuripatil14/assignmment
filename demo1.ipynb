```python
l = [5,4,5,4,6,7,8,4,3,5,3,4]
s = "I love python programming"

# 1. Unique list without changing order
unique_list = []
for x in l:
    if x not in unique_list:
        unique_list.append(x)
print(unique_list)   # [5, 4, 6, 7, 8, 3]
```

```python
# 2. Frequency of every element in a dict
freq_ele = {}
for x in l:
    freq_ele[x] = freq_ele.get(x, 0) + 1
print(freq_ele)      
```

```python
# 3. Frequency of every character in the string
freq_char = {}
for ch in s:
    freq_char[ch] = freq_char.get(ch, 0) + 1
print(freq_char)
```
